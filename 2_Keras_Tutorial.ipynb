```python
# 匯入 Keras 的模型與層：Sequential 代表依序堆疊的網路，Dense 是全連接層
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense

import numpy as np

# 隨機產生 800 筆資料，每筆有 10 個 0~1 的浮點數（特徵）
data = np.random.random((800, 10))
# 隨機產生 800 個標籤，數值只會是 0 或 1（二元分類）
labels = np.random.randint(2, size=(800, 1))

# 檢查資料形狀與第一筆內容
print(np.shape(data))
print(data[0])
print(np.shape(labels))
print(labels[0])

# ===== 建立模型 =====
model = Sequential()
# 隱藏層：16 個神經元，激活函數 relu，輸入維度 10（對應 10 個特徵）
model.add(Dense(16, activation='relu', input_dim=10))
# 輸出層：1 個神經元，sigmoid 輸出 0~1 的機率，適合二元分類
model.add(Dense(1, activation='sigmoid'))

# 編譯模型：指定優化器、損失函數與評估指標
# binary_crossentropy 為二元分類常用的損失函數
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# ===== 訓練模型 =====
# epochs：整份資料重複訓練 15 次；batch_size：每 16 筆資料更新一次權重
model.fit(data, labels, epochs=15, batch_size=16)

# ===== 評估模型 =====
# evaluate 回傳 [loss, accuracy]，score[1] 即為準確率
# 因為資料與標籤都是隨機產生，沒有真實規律，準確率預期接近 0.5
score = model.evaluate(data, labels)
print(score[1])
```
